```python
from diffusers import StableDiffusionImg2ImgPipeline
import torch

# Stable Diffusion v1.5 (InstructPix2Pix)
# Stable Diffusion XL (SDXL - InstructPix2Pix)
# Kandinsky 2.2 (Inpainting 방식)

model_path = "./best_unet_resumed.pt"  # 또는 best_unet.pt
pipe = StableDiffusionImg2ImgPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5",
    torch_dtype=torch.float32
).to("cuda")

pipe.unet.load_state_dict(torch.load(model_path, map_location="cuda"))
pipe.to("cuda")

print("✅ Model loaded successfully!")
```
