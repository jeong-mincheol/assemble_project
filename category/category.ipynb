## 수정

```python
import pandas as pd
import os

# --- 1. 통합할 파일 목록과 결과 파일 이름 정의 ---
category_files = ['area_category.csv', 'cost_category.csv', 'style_category.csv', '카테고리DB.csv']
output_file = '카테고리DB_최종.csv'

# --- 2. 각 카테고리 파일을 읽어서 리스트에 저장 ---
all_category_dfs = []
print("파일을 읽는 중...")
for file in category_files:
    # 파일이 존재하는지 먼저 확인
    if os.path.exists(file):
        try:
            df_temp = pd.read_csv(file)
            # 'code'와 'name' 컬럼이 있는지 확인
            if 'code' in df_temp.columns and 'name' in df_temp.columns:
                all_category_dfs.append(df_temp[['code', 'name']])
                print(f"- '{file}' 파일을 성공적으로 읽었습니다.")
            else:
                print(f"경고: '{file}' 파일에 'code' 또는 'name' 컬럼이 없어 건너뜁니다.")
        except Exception as e:
            print(f"오류: '{file}' 파일을 읽는 중 문제가 발생했습니다: {e}")
    else:
        print(f"경고: '{file}' 파일을 찾을 수 없습니다.")

# --- 3. 데이터프레임 합치기 및 저장 ---
if all_category_dfs:
    # 리스트에 있는 모든 데이터프레임을 하나로 합칩니다.
    df_combined = pd.concat(all_category_dfs, ignore_index=True)

    # 통합된 결과물을 CSV 파일로 저장합니다.
    df_combined.to_csv(output_file, index=False, encoding='utf-8-sig')

    print("-" * 50)
    print(f"성공: {len(all_category_dfs)}개의 카테고리 파일이 '{output_file}' 파일로 통합되었습니다.")
    print("--- 통합된 데이터 상위 5개 ---")
    print(df_combined.head())
    print("-" * 50)
else:
    print("-" * 50)
    print("오류: 통합할 파일이 하나도 없습니다. 파일 이름과 위치를 확인해주세요.")
```
