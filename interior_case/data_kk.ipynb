```python
import pandas as pd

# 예시 df 로드 (사용자 환경에 맞게 변경)
# df = pd.read_csv('master_info.csv')

# 1. 'Space' 컬럼에서 결측값 제외 후 쉼표 기준으로 분리 + 공백 제거
space_split = df['Space'].dropna().apply(lambda x: [s.strip() for s in x.split(',')])

# 2. 리스트를 평탄화(flatten)
all_spaces = [item for sublist in space_split for item in sublist]

# 3. 중복 제거 및 정렬
unique_spaces = sorted(set(all_spaces))

print("유니크 공간 목록 : ", unique_spaces)

# 4. 코드 부여 (기타는 9999, 나머지는 0000부터 4자리 0패딩)
final_area_codes = {}
code = 0
for name in unique_spaces:
    if name == '기타':
        final_area_codes[name] = '9999'
    else:
        final_area_codes[name] = str(code).zfill(4)
        code += 1

# 5. DataFrame 생성 및 번호 순 정렬
df_area = pd.DataFrame([
    {'area_name': name, 'area_no': final_area_codes[name]} for name in final_area_codes
])
df_area = df_area.sort_values(by='area_no').reset_index(drop=True)

# 6. 출력 및 CSV 저장
print(df_area)
df_area.to_csv('area_category.csv', index=False, encoding='utf-8-sig')
```

```python
import pandas as pd
import numpy as np

# --- 1단계: 데이터 로드 및 기본 전처리 ---

# CSV 파일을 읽어서 DataFrame으로 변환
# 'master_info.csv' 파일이 없으면 예시 데이터를 생성합니다.
try:
    df = pd.read_csv('master_info.csv')
except FileNotFoundError:
    print("'master_info.csv' 파일을 찾을 수 없어 예시 데이터로 진행합니다.")
    data = {
        'Seq': [10000, 10001, 10002, 10003, 10004],
        'ShopName': ['한샘', '한샘', '까사미아', '이케아', '한샘'],
        'Address': ['서울', '경기', '서울', '광명', '부산'],
        'Area (평형)': [50, 47, 32, 25, 50],
        'Apt Name': ['래미안', '자이', '힐스테이트', '푸르지오', '래미안'],
        'View Count': [400, 231, 500, 120, 330],
        'Editor': ['에디터A', '에디터B', '에디터A', '에디터C', '에디터B'],
        'Space': ['전체,거실,침실', '거실,키친,욕실', '거실,주방', np.nan, '기타'],
        'Cost': ['7천만원 이상', '5천만원', '5천만원', '3천만원', '7천만원 이상'],
        'Style': ['모던', '클래식&amp;모던', '미니멀', '북유럽', np.nan]
    }
    df = pd.DataFrame(data)


# 불필요한 열 제거
columns_to_drop = ['ShopName', 'Address', 'Area (평형)', 'Apt Name', 'View Count', 'Editor']
existing_columns_to_drop = [col for col in columns_to_drop if col in df.columns]
df.drop(columns=existing_columns_to_drop, inplace=True)

print("--- 1. 초기 데이터 로드 및 불필요한 컬럼 제거 완료 ---")
print(df.head())
print("-" * 50)


# --- 2단계: 'Space' 컬럼 분석 및 카테고리 코드 생성/저장 ---

space_split = df['Space'].dropna().apply(lambda x: [s.strip() for s in x.split(',')])
all_spaces = [item for sublist in space_split for item in sublist]
unique_spaces = sorted(list(set(all_spaces)))

final_area_codes = {}
code = 1
for name in unique_spaces:
    if name == '기타':
        final_area_codes[name] = 'spa_9999'
    else:
        final_area_codes[name] = f"spa_{str(code).zfill(4)}"
        code += 1

df_area = pd.DataFrame(list(final_area_codes.items()), columns=['name', 'code'])
df_area = df_area.sort_values(by='code').reset_index(drop=True)
df_area = df_area[['code', 'name']]
df_area.to_csv('area_category.csv', index=False, encoding='utf-8-sig')

print("--- 2. 'Space' 카테고리 코드 생성 및 'area_category.csv' 저장 완료 ---")
print(df_area)
print("-" * 50)


# --- 3단계: 'Cost' 컬럼 분석 및 카테고리 코드 생성/저장 ---

unique_costs = sorted(df['Cost'].dropna().unique())
final_cost_codes = {}
code = 1
for name in unique_costs:
    final_cost_codes[name] = f"cos_{str(code).zfill(4)}"
    code += 1

df_cost = pd.DataFrame(list(final_cost_codes.items()), columns=['name', 'code'])
df_cost = df_cost.sort_values(by='code').reset_index(drop=True)
df_cost = df_cost[['code', 'name']]
df_cost.to_csv('cost_category.csv', index=False, encoding='utf-8-sig')

print("--- 3. 'Cost' 카테고리 코드 생성 및 'cost_category.csv' 저장 완료 ---")
print(df_cost)
print("-" * 50)


# --- 4단계: 'Style' 컬럼 분석 및 카테고리 코드 생성/저장 ---

# 결측값을 '기타'로 채우고 '&amp;'를 '&'로 변환
df['Style'].fillna('기타', inplace=True)
style_series = df['Style'].str.replace('&amp;', '&', regex=False)

# 쉼표로 분리하고 평탄화하여 유니크한 스타일 목록 추출
style_lists = style_series.str.split(',')
all_styles = [style.strip() for sublist in style_lists for style in sublist]
unique_styles = sorted(list(set(all_styles)))

print("--- 4a. 유니크 스타일 목록 추출 완료 ---")
print(unique_styles)
print("-" * 50)

# 'sty_XXXX' 형식으로 스타일 카테고리 코드 생성
final_style_codes = {}
code = 1
for name in unique_styles:
    if name == '기타':
        final_style_codes[name] = 'sty_9999'
    else:
        final_style_codes[name] = f"sty_{str(code).zfill(4)}"
        code += 1

# 생성된 코드 맵을 DataFrame으로 변환하고 컬럼명/순서 변경
df_style = pd.DataFrame(list(final_style_codes.items()), columns=['name', 'code'])
df_style = df_style.sort_values(by='code').reset_index(drop=True)
df_style = df_style[['code', 'name']]
df_style.to_csv('style_category.csv', index=False, encoding='utf-8-sig')

print("--- 4b. 'Style' 카테고리 코드 생성 및 'style_category.csv' 저장 완료 ---")
print(df_style)
print("-" * 50)


# --- 5단계: 원본 DataFrame에 생성된 코드 적용 ---

# 'Space'를 코드로 변환하는 함수
def convert_space_to_codes(space_string):
    if pd.isna(space_string) or not isinstance(space_string, str):
        return space_string
    spaces = [s.strip() for s in space_string.split(',')]
    codes = [final_area_codes.get(s, "CODE_NOT_FOUND") for s in spaces]
    return ','.join(codes)

# 'Style'을 코드로 변환하는 함수
def convert_style_to_codes(style_string):
    if pd.isna(style_string): # 이미 fillna를 했지만 안전장치로 남겨둠
        return 'sty_9999'
    processed_string = style_string.replace('&amp;', '&')
    styles = [s.strip() for s in processed_string.split(',')]
    codes = [final_style_codes.get(s, "CODE_NOT_FOUND") for s in styles]
    return ','.join(codes)

# 각 컬럼에 코드 적용
df['apply_space'] = df['Space'].apply(convert_space_to_codes)
df['apply_cost'] = df['Cost'].map(final_cost_codes)
df['apply_style'] = df['Style'].apply(convert_style_to_codes)

# 원본 'Space', 'Cost', 'Style' 컬럼은 삭제
df.drop(columns=['Space', 'Cost', 'Style'], inplace=True)


print("--- 5. 원본 데이터에 코드 적용 완료 ---")
# 최종 결과 확인 (apply_ 컬럼들이 추가되고 원본 컬럼 삭제됨)
print(df.head())
print("-" * 50)

# (선택) 모든 전처리가 완료된 DataFrame을 새로운 CSV 파일로 저장
df.to_csv('master_info_preprocessed.csv', index=False, encoding='utf-8-sig')
print("'master_info_preprocessed.csv' 파일 저장이 완료되었습니다.")
```

```python
import pandas as pd

# --- 설정 ---
# 원본 파일과 결과를 저장할 파일 이름을 지정합니다.
input_file = 'text_image_combined.csv'
output_file = 'text_image_deduplicated.csv'

try:
    # --- 1단계: 원본 CSV 파일 로드 ---
    print(f"'{input_file}' 파일을 로드합니다...")
    df = pd.read_csv(input_file)
    rows_before = len(df)
    print(f"-> 중복 제거 전 전체 행의 수: {rows_before}개")
    print("-" * 50)

    # --- 2단계: 중복 행 제거 ---
    # 'subset'에 지정된 컬럼들의 값이 모두 동일한 행을 중복으로 간주합니다.
    # keep='first' 옵션은 중복된 행들 중 첫 번째 행만 남기고 나머지를 삭제합니다.
    print("'Seq', '텍스트' 기준 중복 행을 제거합니다...")
    df_deduplicated = df.drop_duplicates(subset=['Seq', '텍스트'], keep='first')
    rows_after = len(df_deduplicated)
    print("-> 중복 제거 완료!")
    print("-" * 50)

    # --- 3단계: 결과 저장 및 요약 출력 ---
    # 중복이 제거된 결과를 새로운 CSV 파일로 저장합니다.
    df_deduplicated.to_csv(output_file, index=False, encoding='utf-8-sig')

    print("--- 작업 요약 ---")
    print(f"중복 제거 전: {rows_before}개 행")
    print(f"중복 제거 후: {rows_after}개 행")
    print(f"삭제된 중복 행: {rows_before - rows_after}개 행")
    print(f"\n결과가 '{output_file}' 파일로 성공적으로 저장되었습니다.")

except FileNotFoundError:
    print(f"오류: '{input_file}' 파일을 찾을 수 없습니다. 파일이 스크립트와 같은 위치에 있는지 확인해주세요.")
except KeyError as e:
    print(f"오류: {e} 컬럼을 찾을 수 없습니다. '{input_file}' 파일의 컬럼명을 확인해주세요. (예: 'Seq', '텍스트', '이미지')")
except Exception as e:
    print(f"알 수 없는 오류가 발생했습니다: {e}")
```

```python
import pandas as pd

# 삭제 조건 1: 완전 일치하는 문자열 목록
exact_matches = ['▼▼▼', '-', '[공사 중 사진]', '- epilogue -', '# 현장 설명 #', '.']

# 삭제 조건 2: 포함되는 문자열 목록
contains_keywords = [
    '도면', 'before', 'Before', 'BEFORE', '시공 전', '공사 전', '철거', '공사전', '시공전', '안녕', '시 공 전', 
    '이전', '중간', '평면도', '세월의 흔적', 'Bofore', 'plan', '일정표', 'Plan', 'PLAN', '전',
    '준비 중', '작업 중', '진행 중', '시작 전', '완료 전', '진행 상황', '중간 단계',
    '철거 전', '철거 완료', '기초 공사', '해체', '기초', '배관', '배선', '시공 과정', '시공 과정 사진',
    '현장 사진', '작업 사진', '진행 사진', '현장 상황', '현장 기록', '공사 기록',
    '설계도', '플랜', '계획서', '프로젝트 계획', '스케치',
    '전후', '비포', '애프터', 'before & after', 'before-after', '초기', '시작', '시공 전후', '철거 중',
    '요약', '렌더링', 'Rendering', 'rendering', 'after', 'After', 'A안', 'B안',
    '현장 소개', '현장소개', '공사 후', '공사후', 'INTRO', 'Intro', 'intro', '3D'
]

# 조건 1: 완전 일치
cond_exact = df_filtered3['텍스트'].isin(exact_matches)

# 조건 2: 키워드 포함
cond_contains = df_filtered3['텍스트'].str.contains('|'.join(contains_keywords), case=False, na=False)

# 조건 3: 띄어쓰기 없는 텍스트
cond_no_space = df_filtered3['텍스트'].str.strip().str.contains(r'^\S+$', na=False)

# 조건 4: 특수문자로만 구성된 텍스트
cond_special_chars_only = df_filtered3['텍스트'].str.strip().str.match(r'^[^A-Za-z0-9가-힣]+$', na=False)

# ✅ 조건 5: 텍스트가 10글자 이하
cond_short_length = df_filtered3['텍스트'].str.strip().str.len() <= 30

# 조건 11: 텍스트가 큰따옴표로만 감싸진 경우 (ASCII 큰따옴표 " 또는 한글 큰따옴표 “ ”)
cond_quoted_text = df_filtered3['텍스트'].str.strip().str.match(r'^(["“”]).*\1$', na=False)

# 최종 조건 통합
final_condition = (
    cond_exact |
    cond_contains |
    cond_no_space |
    cond_special_chars_only |
    cond_short_length | 
    cond_quoted_text
)

# 삭제할 행 개수 확인
print(f"삭제 대상 행 개수: {final_condition.sum()}")
```

```python
# 조건에 해당하는 행 삭제
df_filtered3 = df_filtered3[~final_condition]
```

## AWS에 이미지 업로드

```python
import boto3
import requests
from urllib.parse import urlparse
import os
from dotenv import load_dotenv

load_dotenv()

# AWS 인증 정보
aws_access_key_id = os.getenv("AWS_ACCESS_KEY_ID")
aws_secret_access_key = os.getenv("AWS_SECRET_ACCESS_KEY")
bucket_name = os.getenv("AWS_S3_BUCKET")
region_name = os.getenv("AWS_REGION")

# S3 클라이언트 생성
s3 = boto3.client(
    's3',
    aws_access_key_id=aws_access_key_id,
    aws_secret_access_key=aws_secret_access_key,
    region_name=region_name
)

# 이미지 URL 리스트
image_urls = [
    "https://example.com/image1.jpg",
    "https://example.com/image2.png",
    # ...
]

for url in image_urls:
    try:
        response = requests.get(url)
        response.raise_for_status()  # 오류 확인
        parsed_url = urlparse(url)
        file_name = os.path.basename(parsed_url.path)

        # S3에 업로드
        s3.put_object(
            Bucket=bucket_name,
            Key=f"built_case_image/{file_name}",
            Body=response.content,
            ContentType=response.headers.get('Content-Type', 'image/jpeg')
        )
        print(f"Uploaded {file_name} to S3.")
    except Exception as e:
        print(f"Failed to upload {url}: {e}")
```
