### 이미지 데이터 개수

```python
import os

# 세 개의 폴더 경로 지정
folders = [
    r"./resized_images/512_resized_interior_images",
    r"./resized_images/512_resized_furniture_images",
    r"./resized_images/512_resized_case_study_images"
]

img_exts = {'.jpg', '.jpeg', '.png', '.bmp', '.webp', '.tiff'}

def count_images_in_folder(folder_path):
    count = 0
    for root, _, files in os.walk(folder_path):
        count += sum(1 for f in files if os.path.splitext(f)[1].lower() in img_exts)
    return count

total = 0
for folder in folders:
    n = count_images_in_folder(folder)
    print(f"{os.path.basename(folder)}: {n}개")
    total += n

print(f"\n총 이미지 개수: {total}개")
```

```python
import os
import matplotlib.pyplot as plt
from matplotlib import font_manager, rc

plt.rcParams['font.family'] = 'Malgun Gothic'  # macOS는 'AppleGothic', Ubuntu는 'NanumGothic'
plt.rcParams['axes.unicode_minus'] = False  # 음수 부호 깨짐 방지

# 폴더 경로 지정
folders = [
    r"./resized_images/512_resized_interior_images",
    r"./resized_images/512_resized_furniture_images",
    r"./resized_images/512_resized_case_study_images"
]

# 이미지 확장자 목록
img_exts = {'.jpg', '.jpeg', '.png', '.bmp', '.webp', '.tiff'}

def count_images_in_folder(folder_path):
    count = 0
    for root, _, files in os.walk(folder_path):
        count += sum(1 for f in files if os.path.splitext(f)[1].lower() in img_exts)
    return count

# 폴더별 개수 계산
folder_names, image_counts = [], []

for folder in folders:
    n = count_images_in_folder(folder)
    clean_name = os.path.basename(folder).replace("512_resized_", "").replace("_images", "")
    folder_names.append(clean_name)
    image_counts.append(n)
    print(f"{clean_name}: {n}개")

print(f"\n총 이미지 개수: {sum(image_counts)}개")


plt.figure(figsize=(8, 5))
bars = plt.bar(folder_names, image_counts, color="#4C72B0")

# 막대 위에 숫자 표시
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, height + 10, f"{height:,}",
             ha='center', va='bottom', fontsize=11, fontweight='bold')

# 그래프 스타일 설정
plt.title("이미지 데이터 분포", fontsize=15, fontweight='bold')
plt.xlabel("데이터 카테고리", fontsize=12)
plt.ylabel("이미지 개수", fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()

# 고해상도(dpi 350)로 저장
plt.savefig("image_distribution.png", dpi=350, bbox_inches='tight')
plt.show()
```

### 시공사례 분포

```python
import os
from collections import defaultdict

folder = "./resized_images/512_resized_case_study_images/"
case_counts = defaultdict(int)

for file_name in os.listdir(folder):
    if not os.path.isfile(os.path.join(folder, file_name)):
        continue
    # 파일명 앞 4자리 추출 (예: 9937_00 → 9937)
    prefix = file_name[:4]
    if prefix.isdigit():  # 숫자로 시작하는 경우만 카운트
        case_counts[prefix] += 1

for case_id, count in sorted(case_counts.items()):
    print(f"시공사례 {case_id}: {count}장")

print(f"\n총 시공사례 수: {len(case_counts)}개")
```

```python
import os
from collections import defaultdict
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

folder = "./resized_images/512_resized_case_study_images/"
case_counts = defaultdict(int)

for file_name in os.listdir(folder):
    if not os.path.isfile(os.path.join(folder, file_name)):
        continue
    prefix = file_name[:4]  # 앞 4자리 추출
    if prefix.isdigit():
        case_counts[prefix] += 1


counts = list(case_counts.values())
print(f"총 시공사례 수: {len(case_counts)}개")
print(f"최소 이미지 수: {min(counts)}장")
print(f"최대 이미지 수: {max(counts)}장")
print(f"평균 이미지 수: {np.mean(counts):.2f}장")

# -------------------------------
# EDA 1: 시공사례별 이미지 개수 히스토그램
# -------------------------------
plt.figure(figsize=(8, 5))
plt.hist(counts, bins=range(1, max(counts) + 2), color="#4C72B0", edgecolor='black', alpha=0.8)
plt.title("시공사례별 이미지 개수 분포", fontsize=15, fontweight='bold')
plt.xlabel("이미지 개수", fontsize=12)
plt.ylabel("시공사례 수", fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.savefig("case_study_image_distribution_hist.png", dpi=350, bbox_inches='tight')
plt.show()

# -------------------------------
# EDA 2: 상위 N개 시공사례 막대그래프 (많은 순)
# -------------------------------
top_n = 15  # 상위 15개만 표시
sorted_cases = sorted(case_counts.items(), key=lambda x: x[1], reverse=True)[:top_n]

plt.figure(figsize=(9, 5))
bars = plt.bar([c[0] for c in sorted_cases], [c[1] for c in sorted_cases], color="#55A868")
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, height + 0.2, str(height),
             ha='center', va='bottom', fontsize=10, fontweight='bold')

plt.title(f"이미지가 많은 상위 {top_n}개 시공사례", fontsize=15, fontweight='bold')
plt.xlabel("시공사례 ID", fontsize=12)
plt.ylabel("이미지 개수", fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.savefig("case_study_image_topN_bar.png", dpi=350, bbox_inches='tight')
plt.show()
```

```python
import os
from collections import defaultdict
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

folder = "./resized_images/512_resized_case_study_images/"
case_counts = defaultdict(int)

for file_name in os.listdir(folder):
    if not os.path.isfile(os.path.join(folder, file_name)):
        continue
    prefix = file_name[:4]
    if prefix.isdigit():
        case_counts[prefix] += 1

case_ids = sorted(case_counts.keys())
counts = [case_counts[cid] for cid in case_ids]

# -------------------------------
# 산점도 (시공사례별 이미지 개수)
# -------------------------------
plt.figure(figsize=(10, 5))
plt.scatter(range(len(case_ids)), counts, color="#4C72B0", alpha=0.7, edgecolor='black')

plt.title("시공사례별 이미지 개수 산점도", fontsize=15, fontweight='bold')
plt.xlabel("시공사례 (ID 순서)", fontsize=12)
plt.ylabel("이미지 개수", fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)

# 통계 기준선 (평균, 최대, 최소)
plt.axhline(np.mean(counts), color='red', linestyle='--', linewidth=1.2, label=f"평균: {np.mean(counts):.2f}")
plt.axhline(max(counts), color='green', linestyle=':', linewidth=1.2, label=f"최대: {max(counts)}")
plt.axhline(min(counts), color='gray', linestyle=':', linewidth=1.2, label=f"최소: {min(counts)}")
plt.legend(fontsize=10)

plt.tight_layout()
plt.savefig("case_study_image_scatter.png", dpi=350, bbox_inches='tight')
plt.show()
```

```python
import os
from collections import defaultdict
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False


folder = r"./interior_case_data/"
case_counts = defaultdict(int)

for file_name in os.listdir(folder):
    if not os.path.isfile(os.path.join(folder, file_name)):
        continue
    prefix = file_name[:4]
    if prefix.isdigit():
        case_counts[prefix] += 1

# -------------------------------
# 기본 통계 계산
# -------------------------------
counts = list(case_counts.values())
mean_val = np.mean(counts)
max_val = max(counts)
min_val = min(counts)

print(f"총 시공사례 수: {len(case_counts)}개")
print(f"평균 이미지 수: {mean_val:.2f}장")
print(f"최소: {min_val}장, 최대: {max_val}장")

# -------------------------------
# 히스토그램 (평균선 추가)
# -------------------------------
plt.figure(figsize=(8, 5))
n, bins, patches = plt.hist(
    counts,
    bins=range(1, max(counts) + 2),
    color="#4C72B0",
    edgecolor='black',
    alpha=0.8
)


plt.axvline(mean_val, color='red', linestyle='--', linewidth=2, label=f"평균: {mean_val:.2f}장")

plt.title("시공사례별 이미지 개수 분포", fontsize=15, fontweight='bold')
plt.xlabel("이미지 개수", fontsize=12)
plt.ylabel("시공사례 수", fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.legend(fontsize=11)
plt.tight_layout()

# 고해상도 저장
plt.savefig("case_study_image_distribution_hist_with_mean.png", dpi=350, bbox_inches='tight')
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os

plt.rcParams['font.family'] = 'NanumGothic'
plt.rcParams['axes.unicode_minus'] = False

# -------------------------------------------------
# CSV 파일 경로 설정
# -------------------------------------------------
folder = r"./interior_case_data/"
master_csv = os.path.join(folder, "master_info_eda.csv")
image_csv = os.path.join(folder, "text_image_combined_eda.csv")

# -------------------------------------------------
# 1️⃣ 시공사례 수 계산 (master_info_eda.csv)
# -------------------------------------------------
master_df = pd.read_csv(master_csv)
unique_cases = master_df['Seq'].nunique()

# -------------------------------------------------
# 2️⃣ 이미지 개수 계산 (text_image_combined_eda.csv)
# -------------------------------------------------
image_df = pd.read_csv(image_csv)

# 각 시공사례별 이미지 수 계산
case_counts = image_df['Seq'].value_counts().to_dict()

# -------------------------------------------------
# 기본 통계 계산
# -------------------------------------------------
counts = list(case_counts.values())
mean_val = np.mean(counts)
max_val = max(counts)
min_val = min(counts)

print(f"총 시공사례 수: {unique_cases}개")
print(f"평균 이미지 수: {mean_val:.2f}장")
print(f"최소: {min_val}장, 최대: {max_val}장")


# -------------------------------
# 히스토그램 (평균선 추가)
# -------------------------------
plt.figure(figsize=(8, 5))
n, bins, patches = plt.hist(
    counts,
    bins=range(1, max(counts) + 2),
    color="#4C72B0",
    edgecolor='black',
    alpha=0.8
)


plt.axvline(mean_val, color='red', linestyle='--', linewidth=2, label=f"평균: {mean_val:.2f}장")

plt.title("시공사례별 이미지 개수 분포", fontsize=15, fontweight='bold')
plt.xlabel("이미지 개수", fontsize=12)
plt.ylabel("시공사례 수", fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.legend(fontsize=11)
plt.tight_layout()

# 고해상도 저장
plt.savefig("case_study_image_distribution_hist_with_mean.png", dpi=350, bbox_inches='tight')
plt.show()
```
